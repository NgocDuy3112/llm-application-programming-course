```python
from openai import OpenAI
from rich import print

client = OpenAI(api_key="none", base_url="http://localhost:11434/v1")
```

```python
response = client.responses.create(model="qwen3:1.7b", input=[{"role": "user", "content": "Hello, in 200 words!"}])
print(response)
```

```python
print(response.output)
```

```python
for block in response.output:
    match block.type:
        case 'reasoning':
            print(f"[bold green]Reasoning:[/bold green] {block.summary[0].text}")
        case 'message':
            print(f"[bold blue]Message:[/bold blue] {block.content[0].text}")
        case _:
            print(f"[bold red]Unknown block type:[/bold red] {block.type}")
```
