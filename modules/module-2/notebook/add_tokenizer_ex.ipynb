## Nhắc lại nhanh — Tại sao cần Tokenizer?

| Vấn đề | Giải pháp |
|--------|-----------|
| Máy tính chỉ hiểu **số**, không hiểu text | Tokenizer chuyển text → dãy số (token IDs) |
| Mỗi model có **vocabulary** riêng | Cùng 1 câu → số token khác nhau tùy model |
| Token count ảnh hưởng đến **chi phí** và **context window** | Cần biết đếm token để kiểm soát budget |

## Bài tập 2: Token Truncation — Xử lý giới hạn Context Window

### 🎯 Mục tiêu
Sau bài tập này, bạn sẽ nắm được:
- Cách đếm token trong text và conversation history bằng `tiktoken`
- Kỹ thuật cắt text (truncation) để không vượt giới hạn context window
- Chiến lược trim conversation history cho chatbot (xóa messages cũ nhất)

### Tại sao cần làm bài này?
Mọi LLM đều có **giới hạn context window**:
- GPT-4o: 128,000 tokens
- Claude 4 Sonnet: 200,000 tokens
- GPT-4o-mini: 128,000 tokens

Nếu input vượt quá giới hạn → **bị lỗi hoặc bị cắt mất thông tin**.

Trong thực tế, khi xây chatbot, conversation history ngày càng dài. Bạn cần biết cách:
1. **Đếm tokens** trong conversation
2. **Cắt bớt** text/conversation khi vượt giới hạn
